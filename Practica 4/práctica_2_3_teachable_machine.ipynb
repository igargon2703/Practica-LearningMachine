````python
import tensorflow as tf
from keras.models import load_model
from PIL import Image, ImageOps
import numpy as np

class_names = open("labels.txt", "r").readlines()
data = np.ndarray(shape=(1, 224, 224, 3), dtype=np.float32)
image = Image.open("incidente.png").convert("RGB")
size = (224, 224)
image = ImageOps.fit(image, size, Image.Resampling.LANCZOS)
image_array = np.asarray(image)
normalized_image_array = (image_array.astype(np.float32) / 127.5) - 1
data[0] = normalized_image_array

prediction = model.predict(data)
index = np.argmax(prediction)
class_name = class_names[index]
confidence_score = prediction[0][index]

# Print prediction and confidence score
print("Class:", class_name[2:], end="")
print("Confidence Score:", confidence_score)
````

**¿Cómo se usa el modelo exportado? ¿Cómo se comporta?**

---



1. El modelo exportado desde Teachable Machine se utiliza cargándolo en Python mediante la librería TensorFlow/Keras (load_model('keras_model.h5')). Para realizar predicciones con una nueva imagen, esta debe cumplir con el preprocesamiento exacto con el que fue entrenado el modelo: abrirse en formato RGB, redimensionarse y recortarse a un tamaño estándar de 224x224 píxeles mediante ImageOps.fit, convertirse en un array de NumPy y normalizar sus valores dividiendo entre 127.5 y restando 1 para acotarlos en el rango $[-1, 1]$. Finalmente, el método model.predict(data) devuelve un vector de probabilidades que se interpreta seleccionando el índice con mayor valor mediante np.argmax.   
2. Con imágenes claras: El modelo muestra un comportamiento robusto, asignando la etiqueta correcta con un nivel de confianza muy alto (cercano al 100% o 1.0).
Con imágenes ambiguas: El modelo tiende a dudar, arrojando porcentajes de confianza más repartidos o moderados (por ejemplo, entre el 40% y el 60%), reflejando la incertidumbre ante características mixtas o mal iluminadas.
Con imágenes fuera de categoría: El modelo fuerza una predicción clasificándola erróneamente en una de las clases existentes, pero por lo general muestra un nivel de confianza notablemente bajo, lo que evidencia que el objeto no encaja adecuadamente en ninguna categoría aprendida.
